```sql
USE CATALOG catalogo_mvp_engenhariadedados;
USE SCHEMA schema_mvp_engenhariadedados;

SELECT  COUNT(*) AS Contagem_linhas,
       COUNT(DISTINCT Student_ID) AS Contagem_distintos,
       COUNT_IF (Student_ID IS NULL) AS Contagem_null,
       COUNT_IF (starting_Salary_USD = 0 ) AS Contagem_semsalario,
       COUNT_IF (starting_Salary_USD > 0 ) AS  Contagem_comsalario,
       COUNT_IF (Placement_Status = 'Placed') AS Contagem_empregados
FROM tabela_estudantes_bronze_v00

```
