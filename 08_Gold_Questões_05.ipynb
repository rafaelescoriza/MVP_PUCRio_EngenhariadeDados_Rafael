```sql
USE CATALOG catalogo_mvp_engenhariadedados;
USE SCHEMA schema_mvp_engenhariadedados;

SELECT 
  ROUND(hrs_estudo_semanal,0) as hrs_estudo_semanal,
  desempenho_academico,
  habil_comunicacao,
  proficiencia_ingles,
  ROUND(AVG(salario_inicial),0) as Salario_Medio
FROM tabela_estudantes_gold_v00
WHERE hrs_estudo_semanal >= 40
  AND (desempenho_academico = 'Excellent' OR desempenho_academico = 'Good')
  AND (proficiencia_ingles = 'Advanced' OR proficiencia_ingles = 'Intermediate')
  AND habil_comunicacao > 7
GROUP BY hrs_estudo_semanal, certificacoes, desempenho_academico, habil_comunicacao, proficiencia_ingles
ORDER BY salario_medio DESC
```
